```python
import cv2

imagem = cv2.imread('dogs.jpeg')

if imagem is None:
    print("Erro ao carregar a imagem")
else:
    dimensoes=imagem.shape
    print(dimensoes)
    ## visualização
    cv2.namedWindow("Dogs", cv2.WINDOW_AUTOSIZE)
    cv2.imshow("Dogs",imagem)
    
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

```python
# carregar video
import cv2

cap= cv2.VideoCapture('cuerdas.mp4')

if not cap.isOpened():
    print("Erro ao carregar vídeo")
else:
    # criando janela
    cv2.namedWindow('janela1', cv2.WINDOW_AUTOSIZE)

    while True:
        sucesso, frame = cap.read()
        if not sucesso:
            print ("Fim do video  ou erro!")
            break
        cv2.imshow('janela1',frame)

        if cv2.waitKey(25) & 0xFF == ord('q'):
            print('encerrado pelo usuario')
            break
    cap.release()
    cv2.destroyAllWindows()
```

```python
# webcan
import cv2
cam = cv2.VideoCapture(0)

if not cam.isOpened():
    print("Erro ao carregar video")
else:
    #Mostra informações
    largura = int(cam.get(cv2.CAP_PROP_FRAME_WIDTH)) 
    altura = int(cam.get(cv2.CAP_PROP_FRAME_HEIGHT))
    fps = cam.get(cv2.CAP_PROP_FPS)
    print("--- INFORMAÇÕES DA WEBCAM ---")
    print(f"Resolução:     {largura} x {altura} pixels")
    print(f"FPS Reportado: {fps}")
    print("Iniciando a janela de visualização. Aperte 'x' para sair.\n")
    # cria janela
    cv2.namedWindow('MyCam', cv2.WINDOW_AUTOSIZE)

    while True:
        sucesso, frame = cam.read()
        frame= cv2.flip(frame, 0)
        # mostra video
        cv2.imshow('MyCam',frame)

        if cv2.waitKey(1) & 0xFF == ord('q'):
            print("encerrado pelo usuario")
            break

    cam.release()
    cv2.destroyAllWindows()
```
